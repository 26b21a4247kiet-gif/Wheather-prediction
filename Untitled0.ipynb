```python
# ============================================================================
# TEAM B6 - MARKOV CHAIN WEATHER PREDICTION
# Project: Steady-State Analysis using Eigenvalue = 1 Method
# Applications: Weather Prediction, Population Dynamics, Queueing Theory
# ============================================================================

import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import eig, matrix_power
import warnings

warnings.filterwarnings("ignore")

# Professional plot style
try:
    plt.style.use("seaborn-v0_8-whitegrid")
except:
    plt.style.use("default")

plt.rcParams["figure.figsize"] = [12, 8]
plt.rcParams["font.size"] = 11

print("=" * 70)
print("TEAM B6 - MARKOV CHAIN ANALYSIS TOOLKIT")
print("=" * 70)


# ============================================================================
# DATASETS
# ============================================================================

BENCHMARK_DATASETS = {

    "weather": {
        "name": "Weather Prediction (3 States)",
        "states": ["Sunny", "Cloudy", "Rainy"],

        # P[i,j] = probability of moving FROM state i TO state j
        "transition_matrix": np.array([
            [0.6, 0.3, 0.1],
            [0.2, 0.5, 0.3],
            [0.1, 0.3, 0.6]
        ]),

        "description":
            "Classic 3-state weather model with moderate persistence"
    },

    "population": {
        "name": "Population Dynamics (Urban/Suburban/Rural)",
        "states": ["Urban", "Suburban", "Rural"],

        "transition_matrix": np.array([
            [0.85, 0.10, 0.05],
            [0.08, 0.82, 0.10],
            [0.12, 0.05, 0.83]
        ]),

        "description":
            "Population migration between residential zones"
    },

    "queueing": {
        "name": "Queueing Theory (Low/Medium/High Load)",
        "states": ["Low Load", "Medium Load", "High Load"],

        "transition_matrix": np.array([
            [0.70, 0.25, 0.05],
            [0.15, 0.60, 0.25],
            [0.05, 0.20, 0.75]
        ]),

        "description":
            "Server load states in queueing systems"
    }
}


# ============================================================================
# LOAD DATASET
# ============================================================================

def load_dataset(dataset_name="weather"):

    if dataset_name not in BENCHMARK_DATASETS:
        raise ValueError(
            f"Dataset '{dataset_name}' not found. "
            f"Choose from: {list(BENCHMARK_DATASETS.keys())}"
        )

    data = BENCHMARK_DATASETS[dataset_name]

    states = data["states"]
    P = data["transition_matrix"].copy()

    print(f"✓ Loaded: {data['name']}")
    print(f"  States: {states}")
    print(f"  {data['description']}")

    return states, P


# ============================================================================
# VALIDATE TRANSITION MATRIX
# ============================================================================

def validate_transition_matrix(P):

    # Check square matrix
    if P.shape[0] != P.shape[1]:
        raise ValueError("Transition matrix must be square.")

    # Check probabilities
    if np.any(P < 0) or np.any(P > 1):
        raise ValueError("Transition probabilities must be between 0 and 1.")

    # Row-stochastic matrix
    row_sums = P.sum(axis=1)

    if not np.allclose(row_sums, 1.0):
        raise ValueError(
            f"Each row must sum to 1. Row sums = {row_sums}"
        )

    return True


# ============================================================================
# BUILD TRANSITION MATRIX
# ============================================================================

def build_transition_matrix(P_dict, states):

    n = len(states)

    P = np.zeros((n, n))

    state_idx = {s: i for i, s in enumerate(states)}

    for (from_s, to_s), prob in P_dict.items():

        i = state_idx[from_s]
        j = state_idx[to_s]

        P[i, j] = prob

    validate_transition_matrix(P)

    return P


# ============================================================================
# STEADY STATE USING EIGENVALUE = 1
# ============================================================================

def find_steady_state_eigenvalue(P, states):

    validate_transition_matrix(P)

    # For row-stochastic P:
    # pi P = pi
    #
    # Therefore pi is the LEFT eigenvector of P.
    #
    # Equivalent:
    # P.T pi.T = pi.T

    eigenvalues, eigenvectors = eig(P.T)

    # Find eigenvalue closest to 1
    idx = np.argmin(np.abs(eigenvalues - 1.0))

    steady_state = np.real(eigenvectors[:, idx])

    # Remove numerical negative signs
    if np.sum(steady_state) < 0:
        steady_state = -steady_state

    # Normalize
    steady_state = steady_state / np.sum(steady_state)

    # Remove tiny numerical errors
    steady_state = np.maximum(steady_state, 0)

    # Normalize again
    steady_state = steady_state / np.sum(steady_state)

    return steady_state


# ============================================================================
# DISTRIBUTION AFTER N DAYS
# ============================================================================

def compute_distribution_after_n_days(P, initial_dist, n_days):

    validate_transition_matrix(P)

    P_n = matrix_power(P, n_days)

    # Row vector formulation:
    # x(n) = x(0) P^n

    final_dist = initial_dist @ P_n

    return final_dist, P_n


# ============================================================================
# VERIFY STEADY STATE
# ============================================================================

def verify_steady_state(P, steady_state, tolerance=1e-6):

    result = steady_state @ P

    error = np.max(
        np.abs(result - steady_state)
    )

    return error < tolerance, error


# ============================================================================
# WEATHER DISTRIBUTION PLOT
# ============================================================================

def plot_weather_distribution(
    states,
    initial_dist,
    steady_state,
    final_dist,
    P,
    n_days=100
):

    fig, axes = plt.subplots(
        1, 2,
        figsize=(14, 6)
    )

    colors = [
        "#FFA726",
        "#42A5F5",
        "#66BB6A"
    ]

    # ------------------------------------------------------------------------
    # PANEL 1
    # ------------------------------------------------------------------------

    x = np.arange(len(states))

    width = 0.25

    axes[0].bar(
        x - width,
        initial_dist,
        width,
        label="Day 0 (Initial)",
        color=colors,
        alpha=0.7
    )

    axes[0].bar(
        x,
        final_dist,
        width,
        label=f"Day {n_days}",
        color=colors,
        alpha=0.7,
        edgecolor="black",
        linewidth=1.5
    )

    axes[0].bar(
        x + width,
        steady_state,
        width,
        label="Steady State (π)",
        color=colors,
        alpha=0.5,
        edgecolor="black"
    )

    axes[0].set_xlabel(
        "Weather State",
        fontsize=12,
        fontweight="bold"
    )

    axes[0].set_ylabel(
        "Probability",
        fontsize=12,
        fontweight="bold"
    )

    axes[0].set_title(
        f"Weather Distribution: Initial vs Day {n_days} vs Steady State",
        fontsize=13,
        fontweight="bold"
    )

    axes[0].set_xticks(x)
    axes[0].set_xticklabels(states)

    axes[0].set_ylim(
        0,
        max(
            initial_dist.max(),
            final_dist.max(),
            steady_state.max()
        ) * 1.3
    )

    axes[0].legend()

    # Value labels
    for i, (init, final, steady) in enumerate(
        zip(initial_dist, final_dist, steady_state)
    ):

        axes[0].text(
            i - width,
            init + 0.02,
            f"{init:.2f}",
            ha="center"
        )

        axes[0].text(
            i,
            final + 0.02,
            f"{final:.2f}",
            ha="center"
        )

        axes[0].text(
            i + width,
            steady + 0.02,
            f"{steady:.2f}",
            ha="center"
        )

    # ------------------------------------------------------------------------
    # PANEL 2 - CONVERGENCE
    # ------------------------------------------------------------------------

    days = [0, 1, 2, 5, 10, 20, 50, 100]

    convergence_data = []

    for day in days:

        if day == 0:
            dist = initial_dist
        else:
            dist, _ = compute_distribution_after_n_days(
                P,
                initial_dist,
                day
            )

        convergence_data.append(dist)

    convergence_data = np.array(convergence_data)

    for i, state in enumerate(states):

        axes[1].plot(
            days,
            convergence_data[:, i],
            marker="o",
            linewidth=2.5,
            label=state,
            color=colors[i],
            markersize=8
        )

        axes[1].axhline(
            y=steady_state[i],
            color=colors[i],
            linestyle="--",
            alpha=0.5
        )

    axes[1].set_xlabel(
        "Days",
        fontsize=12,
        fontweight="bold"
    )

    axes[1].set_ylabel(
        "Probability",
        fontsize=12,
        fontweight="bold"
    )

    axes[1].set_title(
        "Convergence to Steady State",
        fontsize=13,
        fontweight="bold"
    )

    axes[1].set_xticks(days)
    axes[1].set_ylim(0, 1.05)

    axes[1].legend()

    plt.tight_layout()
    plt.show()


# ============================================================================
# SENSITIVITY ANALYSIS
# ============================================================================

def plot_sensitivity_analysis(
    states,
    P_base,
    parameter_name="Sunny→Cloudy",
    variation_range=np.arange(-0.15, 0.16, 0.05)
):

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(14, 6)
    )

    colors = [
        "#FFA726",
        "#42A5F5",
        "#66BB6A"
    ]

    sensitivity_results = []

    base_steady = find_steady_state_eigenvalue(
        P_base,
        states
    )

    # Sunny -> Cloudy
    from_idx = 0
    to_idx = 1

    for delta in variation_range:

        P_modified = P_base.copy()

        new_probability = np.clip(
            P_base[from_idx, to_idx] + delta,
            0,
            1
        )

        # Preserve the other probabilities in Sunny row
        other_indices = [
            i for i in range(len(states))
            if i != to_idx
        ]

        old_other_sum = np.sum(
            P_base[from_idx, other_indices]
        )

        P_modified[from_idx, to_idx] = new_probability

        remaining = 1.0 - new_probability

        if old_other_sum > 0:

            for i in other_indices:

                P_modified[from_idx, i] = (
                    P_base[from_idx, i]
                    / old_other_sum
                    * remaining
                )

        try:

            validate_transition_matrix(P_modified)

            steady_new = find_steady_state_eigenvalue(
                P_modified,
                states
            )

            sensitivity_results.append(
                (delta, steady_new)
            )

        except ValueError:
            continue

    # ------------------------------------------------------------------------
    # Results
    # ------------------------------------------------------------------------

    deltas = np.array(
        [r[0] for r in sensitivity_results]
    )

    steady_states = np.array(
        [r[1] for r in sensitivity_results]
    )

    # ------------------------------------------------------------------------
    # Plot 1
    # ------------------------------------------------------------------------

    for i, state in enumerate(states):

        axes[0].plot(
            deltas,
            steady_states[:, i],
            marker="o",
            linewidth=2.5,
            label=state,
            color=colors[i]
        )

        axes[0].axhline(
            y=base_steady[i],
            color=colors[i],
            linestyle="--",
            alpha=0.5
        )

    axes[0].set_xlabel(
        f"Parameter Variation (Δ{parameter_name})",
        fontsize=12,
        fontweight="bold"
    )

    axes[0].set_ylabel(
        "Steady-State Probability",
        fontsize=12,
        fontweight="bold"
    )

    axes[0].set_title(
        f"Sensitivity Analysis: {parameter_name}",
        fontsize=13,
        fontweight="bold"
    )

    axes[0].legend()
    axes[0].grid(alpha=0.3)

    # ------------------------------------------------------------------------
    # Plot 2
    # ------------------------------------------------------------------------

    if len(sensitivity_results) > 1:

        for i, state in enumerate(states):

            rates = np.gradient(
                steady_states[:, i],
                deltas
            )

            axes[1].plot(
                deltas,
                rates,
                marker="s",
                linewidth=2,
                label=state,
                color=colors[i]
            )

        axes[1].axhline(
            y=0,
            color="black",
            linestyle="-",
            linewidth=1,
            alpha=0.5
        )

        axes[1].set_xlabel(
            f"Parameter Variation (Δ{parameter_name})",
            fontsize=12,
            fontweight="bold"
        )

        axes[1].set_ylabel(
            "Sensitivity (dπ/dparameter)",
            fontsize=12,
            fontweight="bold"
        )

        axes[1].set_title(
            "Rate of Change",
            fontsize=13,
            fontweight="bold"
        )

        axes[1].legend()
        axes[1].grid(alpha=0.3)

    plt.tight_layout()
    plt.show()

    # ------------------------------------------------------------------------
    # Print results
    # ------------------------------------------------------------------------

    print("\n" + "=" * 60)
    print("SENSITIVITY ANALYSIS RESULTS")
    print("=" * 60)

    print(
        "Base steady-state distribution:",
        dict(
            zip(
                states,
                np.round(base_steady, 4)
            )
        )
    )

    print(
        f"Parameter tested: {parameter_name}"
    )

    print(
        f"Variation range: "
        f"{variation_range.min():.2f} "
        f"to "
        f"{variation_range.max():.2f}"
    )

    max_sensitivity = np.max(
        np.abs(
            steady_states - base_steady
        ),
        axis=0
    )

    most_sensitive_idx = np.argmax(
        max_sensitivity
    )

    print(
        f"Most sensitive state: "
        f"{states[most_sensitive_idx]} "
        f"(Δ = "
        f"{max_sensitivity[most_sensitive_idx]:.4f})"
    )

    print("=" * 60)


# ============================================================================
# MAIN PROGRAM
# ============================================================================

print("\n" + "=" * 70)
print("TEAM B6 - MARKOV CHAIN COMPREHENSIVE ANALYSIS")
print("=" * 70 + "\n")


# ============================================================================
# TEST CASE 1 - WEATHER
# ============================================================================

print("📊 TEST CASE 1: WEATHER PREDICTION")
print("-" * 70)

states_weather, P_weather = load_dataset("weather")

# Validate
validate_transition_matrix(P_weather)

# Initial conditions
initial_sunny = np.array([1.0, 0.0, 0.0])
initial_cloudy = np.array([0.0, 1.0, 0.0])
initial_rainy = np.array([0.0, 0.0, 1.0])

# Steady state
steady_weather = find_steady_state_eigenvalue(
    P_weather,
    states_weather
)

print("\n✓ Steady-state distribution:")
for state, prob in zip(
    states_weather,
    steady_weather
):

    print(
        f"  {state}: "
        f"{prob:.4f} "
        f"({prob * 100:.2f}%)"
    )


# Verify
is_valid, error = verify_steady_state(
    P_weather,
    steady_weather
)

print(
    f"\n✓ Verification (π·P = π): "
    f"{'PASS' if is_valid else 'FAIL'} "
    f"(error = {error:.2e})"
)


# 100 days
n_days = 100

final_sunny, _ = compute_distribution_after_n_days(
    P_weather,
    initial_sunny,
    n_days
)

final_cloudy, _ = compute_distribution_after_n_days(
    P_weather,
    initial_cloudy,
    n_days
)

final_rainy, _ = compute_distribution_after_n_days(
    P_weather,
    initial_rainy,
    n_days
)

print(f"\n✓ After {n_days} days:")

print(
    f"  Starting Sunny:  "
    f"{np.round(final_sunny, 6)}"
)

print(
    f"  Starting Cloudy: "
    f"{np.round(final_cloudy, 6)}"
)

print(
    f"  Starting Rainy:  "
    f"{np.round(final_rainy, 6)}"
)


# Convergence check
max_diff = max(
    np.max(
        np.abs(
            final_sunny - steady_weather
        )
    ),
    np.max(
        np.abs(
            final_cloudy - steady_weather
        )
    ),
    np.max(
        np.abs(
            final_rainy - steady_weather
        )
    )
)

print(
    f"\n✓ Convergence check: "
    f"Max deviation = {max_diff:.6f}"
)

print(
    "  → Does it depend on today's weather? "
    + (
        "NO"
        if max_diff < 0.01
        else "YES (not fully converged)"
    )
)


# Plot
plot_weather_distribution(
    states_weather,
    initial_sunny,
    steady_weather,
    final_sunny,
    P_weather,
    n_days
)


# ============================================================================
# TEST CASE 2 - POPULATION
# ============================================================================

print("\n\n📊 TEST CASE 2: POPULATION DYNAMICS")
print("-" * 70)

states_pop, P_pop = load_dataset("population")

steady_pop = find_steady_state_eigenvalue(
    P_pop,
    states_pop
)

print("\n✓ Steady-state population distribution:")

for state, prob in zip(
    states_pop,
    steady_pop
):

    print(
        f"  {state}: "
        f"{prob:.4f} "
        f"({prob * 100:.2f}%)"
    )


# ============================================================================
# TEST CASE 3 - QUEUEING
# ============================================================================

print("\n\n📊 TEST CASE 3: QUEUEING THEORY")
print("-" * 70)

states_queue, P_queue = load_dataset("queueing")

steady_queue = find_steady_state_eigenvalue(
    P_queue,
    states_queue
)

print("\n✓ Steady-state server load distribution:")

for state, prob in zip(
    states_queue,
    steady_queue
):

    print(
        f"  {state}: "
        f"{prob:.4f} "
        f"({prob * 100:.2f}%)"
    )


# ============================================================================
# ANALYSIS COMPLETE
# ============================================================================

print("\n" + "=" * 70)
print("ANALYSIS COMPLETE ✓")
print("=" * 70)


# ============================================================================
# SENSITIVITY ANALYSIS
# ============================================================================

print("\n🔍 PARAMETER SENSITIVITY ANALYSIS")
print("=" * 70)

print(
    "Testing: How does changing "
    "Sunny→Cloudy probability affect steady state?"
)

print("=" * 70 + "\n")


plot_sensitivity_analysis(
    states=states_weather,
    P_base=P_weather,
    parameter_name="Sunny→Cloudy",
    variation_range=np.arange(
        -0.10,
        0.11,
        0.025
    )
)


print("\n💡 Key Insight:")

print(
    "Sensitivity analysis reveals which transition "
    "probabilities have the strongest influence on "
    "long-term behavior."
)

print(
    "This helps identify critical parameters for "
    "system optimization and robustness."
)
```
